# Variables Are Not Permanent

A variable can store a value.

Later in the program, that value can change.

Think of a whiteboard.

You can erase what is written and write something new.

Variables work the same way.

````python
student_score = 45

print(student_score)
````

The variable currently stores the value:

45

Now let's change it.

````python
student_score = 45

print("Original Score:", student_score)

student_score = 78

print("Updated Score:", student_score)
````

Notice something important.

The variable didn't create a second score.

It replaced the old value with the new one.

Variables always contain their **latest assigned value**.

````python
city = "Lagos"

print(city)

city = "Abuja"

print(city)

city = "Port Harcourt"

print(city)
````

Variables are even more powerful because they can work together.

One variable can be used to calculate another.

This allows programs to become dynamic instead of hardcoded.

````python
price = 2500
quantity = 3

total = price * quantity

print(total)
````

Read the code carefully.

The program didn't store the answer manually.

Instead, it calculated it.

This is one of the main reasons we write programs—to automate calculations and decisions.

````python
price = 1800
quantity = 5

total = price * quantity

print("Price:", price)
print("Quantity:", quantity)
print("Total:", total)

print(total)
````

## String Concatenation

### Combining Text in Python

Programs often need to combine multiple pieces of text into one message.

This process is called **string concatenation**.

For example, instead of displaying:

and later displaying:

````python
first_name = "Aisha"

print("Welcome " + first_name)
````

With numbers:

It joins text together.

# A Common Mistake

Look carefully at the following program.

Before running it, try to identify the problem.

````python
age = 20

print("My age is " + age)
````

Did the program produce an error?

Why?

The first part is text.

The second part is a number.

Python does not automatically combine strings and integers.

Let's fix it.

````python
age = 20

print("My age is", age)
````

````python
age = 20

print("My age is " + str(age))
````

Both solutions work.

The first solution separates the values using commas.

The second converts the integer into a string using the `str()` function before joining them together.

You will use both approaches throughout your Python journey.

````python
number = "25"

print(type(number))
````

````python
number = "25"

number = int(number)

print(type(number))
````

Notice what happened.

The value itself still represents twenty-five.

However, its data type has changed.

This allows Python to perform mathematical calculations with it.

````python
first = "15"
second = "8"

print(int(first) + int(second))
````

# Making Our Output Easier to Read

So far, we've displayed information using commas.

For example:

```python
print("Name:", name)
```

This works well.

However, Python provides another way to create cleaner and more readable output.

Let's explore it.

````python
name = "Aisha"
age = 19

print(f"My name is {name} and I am {age} years old.")
````

Notice the letter **f** before the quotation marks.

This tells Python that values inside `{}` should be replaced with the variables.

This is called an **f-string**.

It makes your code easier to read and write.

````python
student = "james"
course = "Artificial Intelligence"

print(f"my name is {student} and I am studying {course}.")
````

### Comparing Different Ways to Print
Python allows us to display variables in different ways.

Let's compare two common approaches.

````python
name = "John"
age = 21

print("Name:", name)
print("Age:", age)

print()

print(f"Name: {name}")
print(f"Age: {age}")
````

Both approaches are correct.

As your programs become larger, you'll often find that **f-strings** are easier to read.

You'll see them frequently in professional Python projects.

## Why Do We Use Loops?

Loops help us when we want to perform the same task multiple times.

Some real-life examples include:

- Printing a student's ID card for every student in a class.
- Calculating the average score for every student.
- Sending reminder emails to multiple customers.
- Processing thousands of records in a dataset.

Instead of writing the same instructions repeatedly, we use a loop.

## Types of Loops in Python

Python provides two main types of loops:

1. **for loop**
2. **while loop**

In this lesson, we'll focus on the **for loop**, which is one of the most commonly used loops in Artificial Intelligence and Data Science.
## What is a `for` Loop?

A **for loop** repeats a block of code for every item in a sequence or for a specific number of times.

### When should you use a `for` loop?

Use a `for` loop when:

- You know how many times you want to repeat a task.
- You want to go through every item in a list, tuple, or dataset.
- You want to process records one after another.

Later in this course, we'll use `for` loops to process datasets and train AI models.

# Introducing the `while` Loop

So far, we've used the **for loop**, which is useful when we know how many times we want to repeat a task.

But what if we **don't know** how many times a task should repeat?

That's where the **while loop** comes in.

## What is a `while` Loop?

A **while loop** repeats a block of code **as long as a condition remains True**.

Unlike a `for` loop, a `while` loop does not require a fixed number of repetitions.

### When should you use a `while` loop?

Use a `while` loop when:

- You don't know how many times the loop should run.
- The loop should continue until a condition changes.
- You want the program to wait until something happens.

Examples include asking a user to enter the correct password or repeating a menu until the user chooses to exit.

````python
count = 1

while count <= 5:
    print("Welcome to AI Class!")
    count = count + 1
````

## Understanding the Code

```python
count = 1
```

Creates a variable named **count** and starts it at 1.

```python
while count <= 5:
```

The loop continues as long as the condition is **True**.

```python
count = count + 1
```

This increases the value of **count** by 1 after each repetition.

If we don't update `count`, the condition will never become **False**, causing the loop to run forever.

## `for` Loop vs `while` Loop

Both loops repeat code, but they are used in different situations.

| `for` Loop | `while` Loop |
|------------|--------------|
| Used when you know the number of repetitions. | Used when the number of repetitions is unknown. |
| Commonly used with `range()`. | Runs until a condition becomes False. |
| Easier to use for counting. | Better for condition-based repetition. |
