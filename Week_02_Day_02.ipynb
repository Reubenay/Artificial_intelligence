## What is NumPy?

**NumPy (Numerical Python)** is a Python library used for fast numerical computation. Its core data structure is the **NumPy array**, which stores numerical data more efficiently than a regular Python list.

Imagine you're analyzing the scores of hundreds of students. Calculating statistics with regular Python lists can become slow and repetitive. NumPy helps you perform these operations with less code and better performance.

````python
import numpy as np

student_scores = np.array([72, 81, 95, 68, 87])

student_scores
````

The output looks similar to a Python list, but it is actually a **NumPy array**.

Let's confirm its data type.

````python
type(student_scores)
````

Expected output:

```python
numpy.ndarray
```

`ndarray` stands for **N-dimensional Array**, the primary data structure provided by NumPy.

## Common NumPy Operations

Creating arrays is just the beginning. NumPy also provides built-in functions for performing quick calculations on numerical data.

Let's use our `student_scores` array to explore some of these operations.

````python
student_scores
````

````python
print("Average Score:", np.mean(student_scores))
print("Highest Score:", np.max(student_scores))
print("Lowest Score:", np.min(student_scores))
print("Total Score:", np.sum(student_scores))
````

These functions help summarize numerical data in just a few lines of code. They are commonly used during data analysis before building Machine Learning models.

````python
# Add 5 bonus marks to every student
updated_scores = student_scores + 5

updated_scores
````

NumPy performs operations on every element in an array at once. This makes numerical computations simple, readable and efficient.

# Data Visualization with Matplotlib

Looking at numbers alone can make it difficult to identify patterns. Data visualization helps us communicate information using charts and graphs.

**Matplotlib** is one of the most popular Python libraries for creating visualizations.

````python
import matplotlib.pyplot as plt
````

## What is `plt.figure()`?

Before drawing a chart, Matplotlib creates a **figure**.

A **figure** is the area or canvas where the chart will be displayed.

The `figsize` argument controls the width and height of the figure.

```python
plt.figure(figsize=(8,5))
```

- **8** → Width
- **5** → Height

## Creating Your First Line Chart

Before creating a chart, let's understand some common Matplotlib functions we'll use:

- **`plt.title()`** – Adds a title to the chart.
- **`plt.xlabel()`** – Labels the horizontal (x-axis).
- **`plt.ylabel()`** – Labels the vertical (y-axis).
- **`plt.grid(True)`** – Displays grid lines to make the chart easier to read.
- **`plt.show()`** – Displays the completed chart.

Let's see these functions in action.

````python
plt.figure(figsize=(8,4))

plt.plot(df["Final_Score"].head(), marker="o")

plt.title("Final Scores of First 10 Students")
plt.xlabel("Student")
plt.ylabel("Final Score")

plt.grid()

plt.show()
````

Notice how the line rises and falls as the scores change.

Line charts are commonly used to visualize trends over time or across ordered observations.

### What Do You Notice?

- The title describes what the chart represents.
- The **x-axis** shows the student positions (first to tenth student).
- The **y-axis** shows the corresponding final scores.
- The markers (`o`) make each student's score easier to identify.
- The grid lines help us estimate the values more accurately.

## Creating a Bar Chart

A **bar chart** is used to compare values across different categories.

In this example, each bar represents a department, while the height of the bar represents the department's average final score.

````python
department_scores = df.groupby("Department")["Final_Score"].mean()

department_scores
````

````python
plt.figure(figsize=(8,5))

department_scores.plot(kind="bar")

plt.title("Average Final Score by Department")
plt.xlabel("Department")
plt.ylabel("Average Final Score")

plt.show()
````

### Observation

- Each bar represents one department.
- Taller bars indicate higher average scores.
- Shorter bars indicate lower average scores.

Bar charts make it easy to compare categories at a glance.

From the chart, we can quickly compare how each department performed on average.

This is much easier than comparing rows of numbers.

## Creating a Histogram

A **histogram** shows how values are distributed. Instead of comparing categories, it groups similar values into ranges (called bins).

Histograms are useful for answering questions like:

- Are most students scoring high or low?
- Are the scores evenly distributed?
- Are there any unusual values?

````python
plt.figure(figsize=(8,5))

plt.hist(df["Final_Score"], bins=10)

plt.title("Distribution of Final Scores")
plt.xlabel("Final Score")
plt.ylabel("Number of Students")

plt.show()
````

### Observation

Each bar represents a range of final scores.

The height of a bar tells us how many students fall within that score range.

Histograms help us understand the distribution of numerical data.
