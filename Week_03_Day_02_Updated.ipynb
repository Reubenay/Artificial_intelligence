````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

## What is Seaborn?

**Seaborn is a Python library built on top of Matplotlib that makes statistical data visualization easier.**

Matplotlib gives us detailed control over charts, while Seaborn provides simpler functions for creating many common statistical visualizations.

We can use both libraries together.


````python
df = pd.read_csv("student_performance_cleaned.csv")

df.head()
````

````python
# Quick Matplotlib recap

plt.figure(figsize=(8, 5))

plt.bar(
    df["Department"].value_counts().index,
    df["Department"].value_counts().values
)

plt.title("Number of Students by Department")
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.xticks(rotation=15)
plt.show()
````

## What is a Scatter Plot?

A **scatter plot** shows the relationship between two numerical variables.

Each point represents one record in the dataset.

For example, we can check whether students who study more tend to score higher in their exams.


## Scatter Plot

A scatter plot displays the relationship between two numerical variables.

Each dot represents one record in the dataset.

### Real-life Example

Suppose you want to know whether students who study longer usually obtain better scores.

Instead of comparing 50 rows in a table, a scatter plot allows you to see the relationship instantly.

````python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Final_Score"
)

plt.title("Study Hours vs Final Score")

plt.show()
````

### Understanding the Code

- **sns** refers to the Seaborn library.
- **scatterplot()** creates a scatter plot.
- **x** determines what appears on the horizontal axis.
- **y** determines what appears on the vertical axis.

Each point represents one student.

## Interpretation

If the points generally move upward from left to right, it suggests that students who study more tend to achieve higher scores.

If the points appear randomly scattered, there may be little or no relationship.

## Count Plot

A count plot shows how many observations belong to each category.

### Real-life Example

Suppose the school principal asks:

> "Which department has the highest number of students?"

Rather than counting manually, a count plot answers the question immediately.

````python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Department"
)

plt.title("Number of Students by Department")

plt.xticks(rotation=15)

plt.show()
````

### Understanding the Code

- **countplot()** counts the number of records in each category.
- Each bar represents one department.
- Taller bars indicate more students.

## Histogram with Seaborn

Earlier, we created a histogram using Matplotlib.

Seaborn provides **histplot()**, which creates a cleaner histogram with less code.

## Understanding Bins

A histogram groups values into intervals called **bins**.

For example,

Scores

60
61
62
63
64

may all belong to the same bin.

Instead of drawing one bar for every score, the histogram groups similar values together.

````python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="Final_Score",
    bins=10
)

plt.title("Distribution of Final Scores")

plt.show()
````

## Understanding a Box Plot

A **box plot** helps us understand the distribution and spread of numerical data and can help us identify possible outliers.

A box plot summarizes data using:

- Minimum
- First Quartile (Q1)
- Median
- Third Quartile (Q3)
- Maximum

Points beyond the whiskers may represent unusually high or low values.


````python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    y="Study_Hours"
)

plt.title("Study Hours Distribution")

plt.show()
````

````python
correlation = df.corr(numeric_only=True)

correlation
````

````python
plt.figure(figsize=(10,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="Blues"
)

plt.title("Correlation Heatmap")

plt.show()
````

## Understanding Correlation

**Correlation measures how strongly two numerical variables move together.**

For example:

- Do students who study longer tend to have higher scores?
- Do students with higher attendance tend to have higher final scores?

Correlation is useful when exploring relationships between numerical variables.

The correlation value ranges from **-1 to 1**.


### How to Read Correlation

| Correlation | Meaning |
|---|---|
| Close to **+1** | Strong positive relationship |
| Close to **0** | Little or no linear relationship |
| Close to **-1** | Strong negative relationship |

A value closer to +1 or -1 indicates a stronger linear relationship.

**Important:** Correlation shows association, but it does not prove that one variable causes the other.


## The `corr()` Function

Pandas provides the `corr()` function to calculate correlations between numerical columns.

We can first select the numerical columns in our dataset and then calculate their correlations.


````python
correlation = df.select_dtypes(include="number").corr()

correlation
````

### Understanding the Code

- `select_dtypes(include="number")` selects the numerical columns.
- `.corr()` calculates the correlation between those numerical columns.
- The result is called a **correlation matrix** because it compares each numerical variable with the others.


## Visualizing Correlation with a Heatmap

A **heatmap** uses colour to represent values in a table.

For correlation analysis, it helps us quickly see which numerical variables have stronger or weaker relationships.


````python
plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="Blues"
)

plt.title("Correlation Heatmap")
plt.show()
````

### Understanding the Heatmap Code

- `correlation` → the correlation matrix we want to visualize.
- `annot=True` → displays the correlation values inside the cells.
- `cmap="Blues"` → controls the colour palette used in the heatmap.
- `plt.figure()` → controls the size of the chart.
- `plt.title()` → gives the chart a title.


### Observation

Look at the heatmap and identify which variables have the strongest positive or negative relationships with `Final_Score`.

Remember: a strong correlation can help us identify useful features, but it does not prove causation.


## Bonus: Pair Plot

A **pair plot** compares several numerical variables at the same time.

It automatically creates multiple scatter plots and distributions, allowing us to quickly look for patterns across several variables.

You do not need to memorize this function yet. The goal is to understand what information the visualization provides.


````python
sns.pairplot(
    df[
        [
            "Attendance",
            "Study_Hours",
            "Assignment_Score",
            "Final_Score"
        ]
    ]
)

plt.show()
````
