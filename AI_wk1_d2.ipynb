# Traditional Programming
Before Machine Learning existed, programmers wrote rules for almost everything.

For example,

If a student's score is greater than or equal to 50,

the student passes.

Otherwise,

Let's implement that rule.

````python
score = int(input("Enter student's score: "))

if score >= 50:
    print("PASS")
else:
    print("FAIL")
````

# Another Example

Imagine you receive 10,000 emails every day.

Some are spam.

Some are important.

Could you write rules for every spam email?

For example,

```
If the email contains the word "Congratulations", it is spam.
```

Would that always work?

No.

Many legitimate emails also contain the word "Congratulations."

We need something smarter.

We need a system that learns from many examples instead of relying on a few hand-written rules.

````python
emails = [
    "Congratulations! You won a prize!",
    "Meeting starts by 9 AM.",
    "Limited offer! Buy now!",
    "Your assignment has been graded.",
    "Click here to claim your reward."
]

for email in emails:
    print(email)
````

Run the program.

Look carefully at the emails.

Without using a computer...

Which ones do you think are spam?

Why?

Notice that you naturally looked for **patterns**.

That is exactly what Machine Learning tries to do with data.

````python
spam_keywords = [
    "won",
    "prize",
    "offer",
    "reward",
    "claim"
]

print("Spam keywords")

for word in spam_keywords:
    print("-", word)
````

Imagine writing rules for every spam keyword.

Today you know:

- prize

- offer

Your program would quickly become outdated.

Machine Learning helps because it can discover patterns from data instead of relying only on manually written rules.

# Machine Learning

Machine Learning is a branch of Artificial Intelligence that enables computers to learn patterns from data instead of relying solely on explicitly programmed rules.

Think back to the spam email example.

Instead of telling the computer every possible spam rule, we provide many examples of:

- Spam emails
- Legitimate emails

The computer studies these examples and learns patterns that help it make predictions on new emails.

Think of a child learning to identify fruits.

You don't explain every possible feature of an apple.

Instead, you show many apples.

🍎 Apple

🍎 Apple

🍎 Apple

Eventually, the child recognizes new apples without being told every rule.

Machine Learning works in a similar way.

It learns from examples.

````python
student_scores = [
    45,
    52,
    61,
    75,
    80,
    93
]

print(student_scores)
````

Imagine these scores belong to previous students.

A teacher can quickly observe that most students scored above 50.

Humans naturally identify patterns.

Machine Learning attempts to identify useful patterns automatically from much larger datasets.

````python
student_scores = [45, 52, 61, 75, 80, 93,]

average = sum(student_scores) / len(student_scores)

print("Average Score:", average)
````

Run the program.

Run it again.

Questions to consider:

- Did the average change?
- Why did it change?
- What happens if you add a very low score?

Exploring data like this is often the first step before building Machine Learning models.

````python
highest_score = max(student_scores)
lowest_score = min(student_scores)

print("Highest:", highest_score)
print("Lowest:", lowest_score)
````

These are simple calculations.

They are **not Machine Learning**.

# How Do Machines Learn From Data?
## From Data to Decisions

We've learned that Machine Learning learns from data.

But a question naturally follows.

How?

How does a computer look at thousands—or even millions—of examples and begin making predictions?

The answer is that it looks for **patterns**.

It doesn't understand the data the way humans do.

Instead, it finds mathematical relationships that help it make future predictions.

Let's explore this idea with a few simple examples.

````python
hours_studied = [1, 2, 3, 4, 5]
exam_scores = [35, 48, 60, 73, 88]

for hour, score in zip(hours_studied, exam_scores):
    print(f"Hours Studied: {hour} --> Exam Score: {score}")
````

Look carefully at the data.

As the number of hours studied increases...

What happens to the scores?

Most people immediately notice a pattern.

The computer can also discover relationships like this when given enough examples.

This ability to recognize patterns is at the heart of Machine Learning.

````python
hours = [1, 2, 3, 4, 5]
scores = [35, 48, 60, 73, 88]

print("Highest Score:", max(scores))
print("Lowest Score:", min(scores))

difference = max(scores) - min(scores)

print("Difference:", difference)
````

## NOTE
This program isn't performing Machine Learning.

It is helping us understand the data.

Before data scientists build models, they spend time exploring their datasets.

This process is called **Exploratory Data Analysis (EDA)**.

We'll learn much more about EDA in later weeks.

# Good Data vs Bad Data
Imagine you want to teach a computer to recognize dogs.

Dataset A contains:

- Beagle

Dataset B contains:

- Labrador
- Labrador
- Labrador

Which dataset would help the computer learn better?

Probably Dataset A.

Why?

Because it contains more variety.

````python
dataset_a = [
    "Labrador",
    "German Shepherd",
    "Poodle",
    "Bulldog",
    "Beagle"
]

dataset_b = [
    "Labrador",
    "Labrador",
    "Labrador",
    "Labrador",
    "Labrador"
]

print("Dataset A:", len(dataset_a), "examples")
print("Dataset B:", len(dataset_b), "examples")
````

Even though both datasets contain five examples, they are not equally useful.

A diverse dataset often helps Machine Learning models perform better because it exposes them to different kinds of examples.

Remember:

More data is helpful.

Better data is even more important.

````python
student_data = [
    {"name": "Ada", "score": 75},
    {"name": "James", "score": 62},
    {"name": "Fatima", "score": 90},
    {"name": "David", "score": 54},
    {"name": "Grace", "score": 81}
]

for student in student_data:
    print(student)
````

Instead of looking at just numbers, we're now working with structured data.

Each student has:

- A name

This is closer to the kind of data you'll work with in real AI projects.

In later weeks, you'll learn how libraries like **Pandas** make working with this kind of data much easier.

````python
scores = []

for student in student_data:
    scores.append(student["score"])

print("Scores:", scores)
print("Average:", sum(scores) / len(scores))
````

# Step 1 — Define the Problem

Every AI project starts with a problem.

Examples include:

- Detect spam emails.
- Predict house prices.
- Recommend movies.
- Translate languages.
- Detect fraudulent bank transactions.

Without a clear problem, there is nothing for AI to solve.

# Step 2 — Collect Data

After defining the problem, the next step is collecting data.

Examples include:

For spam detection:

- Thousands of spam emails
- Thousands of legitimate emails

For predicting house prices:

- House size
- Location
- Number of bedrooms
- Selling price

Good AI begins with good data.

````python
houses = [
    {"bedrooms": 2, "price": 150000},
    {"bedrooms": 3, "price": 240000},
    {"bedrooms": 4, "price": 350000},
]

for house in houses:
    print(house)
````

This small dataset contains only three houses.

Real Machine Learning projects may contain thousands—or even millions—of records.

We'll learn how to work with larger datasets in later weeks using **Pandas**.

# Step 3 — Learn from the Data

Once data has been collected, Machine Learning algorithms begin searching for patterns.

For example,

The algorithm may discover that:

- Larger houses often cost more.
- Houses in certain locations are more expensive.
- More bedrooms generally increase the selling price.

Notice something important.

Nobody explicitly wrote these rules.

The algorithm discovers them from the data.

# Step 4 — Make Predictions

Once the model has learned from previous examples,

it can begin making predictions for new examples.

For instance,

A new house has:

- 3 Bedrooms

The model estimates the likely selling price based on what it learned from previous houses.

This ability to generalize to new examples is one of the defining characteristics of Machine Learning.

# Step 5 — Evaluate and Improve

Building an AI system doesn't end after making predictions.

Developers ask questions like:

- Is the prediction accurate?
- Can the model improve?
- Do we need more data?
- Is the data balanced?
- Should we try another algorithm?

Developing AI is an iterative process.

### Finding Patterns
Machine Learning is interested in finding patterns.

Let's look for some patterns ourselves before asking a computer to do it.

Study the following data carefully.

````python
sales = [
    {"month": "January", "sales": 120},
    {"month": "February", "sales": 150},
    {"month": "March", "sales": 180},
    {"month": "April", "sales": 240},
    {"month": "May", "sales": 310}
]

for month in sales:
    print(month)
````

````python
highest_month = sales[0]

for month in sales:
    if month["sales"] > highest_month["sales"]:
        highest_month = month

print(highest_month)
````

This program finds the month with the highest sales.

Again,

this isn't Machine Learning.

It is data processing.

Learning to process data is an essential skill because Machine Learning models depend on clean, well-understood data.
