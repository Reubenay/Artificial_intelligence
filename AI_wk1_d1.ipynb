# Artificial Intelligence vs Automation

Many people use these two terms interchangeably.

However, they are not the same.

Automation simply means a machine or software performs the same task repeatedly according to predefined instructions.

Artificial Intelligence goes a step further.

Instead of only following fixed instructions, AI can learn from data and improve its performance over time.

### Example 1

Imagine your phone alarm rings every morning at exactly 6:00 AM.

Is this Artificial Intelligence?

No.

The alarm simply follows a schedule.

That is **automation**.

````python
alarm_time = "06:00"

current_time = input("Enter the current time (HH:MM): ")

if current_time == alarm_time:
    print("Wake up!")
else:
    print("Keep sleeping.")
````

This program follows one simple rule.

It doesn't learn.

It doesn't adapt.

It doesn't improve.

It simply executes instructions.

This is automation.

### Example 2

Now imagine Spotify.

It doesn't recommend the same songs to everyone.

Instead, it learns from:

- Songs you skip
- Songs you replay
- Your favourite artists
- Your listening habits

This is an example of Artificial Intelligence because the recommendations improve based on data.
